# **Data**
---

Source tables, loaded from the CSV paths defined in `utils/paths.py`.

| Table | File | Description |
|---|---|---|
| `orders` | `olist_orders_dataset.csv` | Orders, with status and timestamps |
| `order_items` | `olist_order_items_dataset.csv` | Order lines, at product level |
| `payments` | `olist_order_payments_dataset.csv` | Payments for each order |
| `reviews` | `olist_order_reviews_dataset.csv` | Customer reviews for each order |
| `customers` | `olist_customers_dataset.csv` | Customer records linked to each order |
| `products` | `olist_products_dataset.csv` | Product catalog |
| `sellers` | `olist_sellers_dataset.csv` | Seller records |
| `geolocation` | `olist_geolocation_dataset.csv` | Coordinates by zip code prefix |
| `category_translation` | `product_category_name_translation.csv` | Product category names in Portuguese and English |

**orders**

```python
data = pd.read_csv(ORDERS_FILE)
```

```python
dataset_summary(data)
```

```python
dataset_columns(data)
```

- `orders` has 99,441 rows and 8 columns. Null fields appear only in the fulfillment timestamps: `order_approved_at` (160, 0.16%), `order_delivered_carrier_date` (1,783, 1.79%) and, most of all, `order_delivered_customer_date` (2,965, 2.98%).
- The timestamp columns are read as strings (`object`) and need to be converted to datetime before any time or delivery analysis.
