# **Uniqueness**
---

`order_id` and `customer_id` are unique, so `orders` has one row per order and each order carries its own `customer_id`. `order_status` (8 values) and `order_estimated_delivery_date` (459 distinct dates) repeat as expected for low-cardinality columns.

```python
evaluate_uniqueness(orders_items)
```

`order_id` repeats in 87.59% of the rows because an order can have several items, which is the expected one-to-many relationship with `orders`. `product_id` and `seller_id` also repeat, since the same product or seller appears in many lines.

```python
evaluate_uniqueness(payments_file)
```

`order_id` is duplicated in 4.28% of the rows, which fits orders paid with more than one method or in several parts (`payment_sequential` reaches 29). `payment_type` has 5 distinct values.

```python
evaluate_uniqueness(reviews_file)
```

`review_id` is duplicated in 0.82% of the rows and `order_id` in 0.56%, so a small number of orders have more than one review. `review_score` has the 5 expected values (1 to 5).

```python
evaluate_uniqueness(customers_file)
```

`customer_id` is unique, so it identifies an order-level customer record. `customer_unique_id` repeats in 3.36% of the rows: those are people who ordered more than once under different `customer_id` values. Any repeat-purchase analysis has to use `customer_unique_id`.

```python
evaluate_uniqueness(products_file)
```

`product_id` is unique, so `products` has one row per product. The other columns are low-cardinality descriptive attributes (74 categories, dimension and weight values), so their repetition is expected.

```python
evaluate_uniqueness(sellers_file)
```

`seller_id` is unique. `seller_zip_code_prefix` repeats in 27.43% of the rows, and `seller_city` and `seller_state` repeat even more because many sellers share a location.

```python
evaluate_uniqueness(geolocation_file)
```

No column in `geolocation` is unique, and `geolocation_zip_code_prefix` repeats in 98.10% of the rows. The table stores many latitude/longitude points per zip code prefix, so the repetition is by design.

```python
evaluate_uniqueness(category_translation_file)
```

Both columns are unique, so each Portuguese category name maps to exactly one English name.

# **Distribution**
---

```python
evaluate_distribution(orders)
```

`order_status` is dominated by `delivered` (97.02%), so cancellations and other non-delivered states are a small share of the table. This matters for delivery-time analysis. The five timestamps now load as datetimes: purchases run from 2016-09-04 to 2018-10-17 (about 25 months) and estimated delivery dates go up to 2018-11-12. The first and last months are incomplete, so period comparisons need care. The identifier columns are all unique, so their top value says nothing.

```python
evaluate_distribution(orders_items)
```

`price` ranges from 0.85 to 6,735.00, with a mean of 120.65 against a median of 74.99, which points to a right-skewed distribution with high-value outliers. `freight_value` behaves the same way (mean 19.99, median 16.26). `order_item_id` is the position of the line within the order (maximum 21), not a quantity. There are 112,650 lines for 98,666 orders, about 1.14 items per order, and only 9.9% of orders have more than one item. `shipping_limit_date` falls inside the 2016-2018 window except for 4 lines (3 distinct timestamps) dated in 2020, which look like errors and should be excluded from time-based analysis.

```python
evaluate_distribution(payments_file)
```

`credit_card` is the most used `payment_type` (73.92%). `payment_value` is strongly right-skewed (mean 154.10, median 100.00, maximum 13,664.08). `payment_installments` averages under 3, with a maximum of 24.

```python
evaluate_distribution(reviews_file)
```

`review_score` has a mean of 4.09 and a median of 5.0, so scores lean strongly positive. Together with the low completeness of the comment fields, this suggests most reviews are quick ratings without a written comment. Review creation dates run from 2016-10-02 to 2018-08-31, and answer timestamps go up to 2018-10-29.

```python
evaluate_distribution(customers_file)
```

`customer_state` is concentrated in `SP` (41.98% of customers), and `sao paulo` is the most common city (15.63%). `customer_zip_code_prefix` spans the whole national range (1,003 to 99,990).

```python
evaluate_distribution(products_file)
```

`cama_mesa_banho` (bed, bath and table) is the largest category, with 9.37% of the products. `product_weight_g` is strongly right-skewed (mean 2,276.47, median 700, maximum 40,425), so a few heavy items pull the mean well above the typical product. Length, height and width are less spread out.

```python
evaluate_distribution(sellers_file)
```

`seller_state` is even more concentrated than `customer_state`: `SP` holds 59.74% of the sellers and `sao paulo` is the top city (22.42%). Sellers are clearly more clustered than customers.

```python
evaluate_distribution(geolocation_file)
```

Latitude and longitude range from about -36 to 45 and from -101 to 121, well outside Brazil. Using a bounding box for Brazil (latitude -33.75 to 5.27, longitude -73.99 to -34.79), 42 rows (0.004%) fall outside it. These are coordinate errors and should be filtered before any geospatial work. `SP` is again the top state (40.42%).

```python
evaluate_distribution(category_translation_file)
```

With 71 categories and the most frequent one at only 1.41%, `category_translation` is a flat lookup table with no dominant value.

# **Referential integrity**
---

```python
untranslated = set(products_file["product_category_name"].dropna()) - set(
    category_translation_file["product_category_name"]
)

referential = pd.DataFrame(
    {
        "Records": {
            "orders without order_items": ~orders["order_id"].isin(orders_items["order_id"]),
            "order_items without orders": ~orders_items["order_id"].isin(orders["order_id"]),
            "payments without orders": ~payments_file["order_id"].isin(orders["order_id"]),
            "reviews without orders": ~reviews_file["order_id"].isin(orders["order_id"]),
            "orders without customers": ~orders["customer_id"].isin(customers_file["customer_id"]),
            "order_items without products": ~orders_items["product_id"].isin(products_file["product_id"]),
            "order_items without sellers": ~orders_items["seller_id"].isin(sellers_file["seller_id"]),
        }
    }
).apply(lambda column: column.map(lambda mask: int(mask.sum())))

referential.loc["product categories without translation"] = len(untranslated)
referential
```

```python
orders.loc[~orders["order_id"].isin(orders_items["order_id"]), "order_status"].value_counts()
```

```python
sorted(untranslated)
```

There are no orphan records: every `order_id` in `order_items`, `payments` and `reviews` exists in `orders`, and every customer, product and seller referenced exists in its table. Two gaps affect the KPIs:

- 775 orders have no items (603 `unavailable`, 164 `canceled`, and 8 in other statuses). They count as orders but add no GMV, so the status filter has to be decided before counting orders or computing GMV.
- 2 product categories (`pc_gamer` and `portateis_cozinha_e_preparadores_de_alimentos`) are not in the translation table, so an inner join would drop those products.

# **Consistency**
---

```python
delivered = orders["order_status"] == "delivered"

consistency = pd.DataFrame(
    {
        "Records": {
            "delivered without delivery date": delivered & orders["order_delivered_customer_date"].isna(),
            "delivered without approval date": delivered & orders["order_approved_at"].isna(),
            "not delivered with delivery date": ~delivered & orders["order_delivered_customer_date"].notna(),
            "delivery before purchase": orders["order_delivered_customer_date"] < orders["order_purchase_timestamp"],
        }
    }
).apply(lambda column: column.map(lambda mask: int(mask.sum())))

consistency
```

Order status and dates disagree in a few orders: 8 `delivered` orders have no delivery date, 14 have no approval date, and 6 orders that were not delivered do have a delivery date. No delivery happens before its purchase. Delivery-time KPIs should only use orders that have both dates.

# **Validity**
---

```python
validity = pd.Series(
    {
        "payments with value 0": int((payments_file["payment_value"] == 0).sum()),
        "payments with type not_defined": int((payments_file["payment_type"] == "not_defined").sum()),
        "order_items with shipping_limit_date after 2018": int(
            (orders_items["shipping_limit_date"] > "2018-12-31").sum()
        ),
        "geolocation points outside Brazil": int(
            (
                ~(
                    geolocation_file["geolocation_lat"].between(-33.75, 5.27)
                    & geolocation_file["geolocation_lng"].between(-73.99, -34.79)
                )
            ).sum()
        ),
    },
    name="Records",
).to_frame()

validity
```

```python
payments_file.loc[payments_file["payment_value"] == 0, "payment_type"].value_counts()
```

Some values fall outside what is expected:

- 9 payments have a value of 0 (6 `voucher`, 3 `not_defined`), and 3 payments have the type `not_defined`.
- 4 order items have a shipping limit date in 2020, outside the period covered by the data.
- 42 geolocation points lie outside the bounding box of Brazil (latitude -33.75 to 5.27, longitude -73.99 to -34.79).
