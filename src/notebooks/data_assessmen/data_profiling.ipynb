```python
# Utils
# =============================================================================
from utils.paths import (
    ORDERS_FILE, 
    ORDER_ITEMS_FILE, 
    PAYMENTS_FILE, 
    REVIEWS_FILE, 
    CUSTOMERS_FILE, 
    PRODUCTS_FILE, 
    SELLERS_FILE, 
    GEOLOCATION_FILE, 
    CATEGORY_TRANSLATION_FILE
)

# DATA
# =============================================================================
import pandas as pd

# EDA
# =============================================================================
from data_assessmen.data_profiling import (
    evaluate_completeness,
    evaluate_uniqueness,
    evaluate_distribution)
```

# **Data**
---
**`OLIST_FILES`**

A dictionary that maps each table's short internal name (used throughout the code) to its actual CSV filename as distributed by Kaggle.

Keeping this mapping in one dictionary avoids hardcoding filenames in multiple places, and lets the loader iterate over all tables with a single loop instead of repeating near-identical code nine times.

| Key | File | Description |
|---|---|---|
| `orders` | `olist_orders_dataset.csv` | Core orders table; source of order status and timestamps |
| `order_items` | `olist_order_items_dataset.csv` | Line items linked to each order, at product level |
| `payments` | `olist_order_payments_dataset.csv` | Payment records linked to each order |
| `reviews` | `olist_order_reviews_dataset.csv` | Customer reviews linked to each order |
| `customers` | `olist_customers_dataset.csv` | Customer records linked to each order |
| `products` | `olist_products_dataset.csv` | Product catalog linked to order items |
| `sellers` | `olist_sellers_dataset.csv` | Seller records linked to order items |
| `geolocation` | `olist_geolocation_dataset.csv` | Reference table of zip-code coordinates |
| `category_translation` | `product_category_name_translation.csv` | Reference table translating product categories |

```python
orders = pd.read_csv(ORDERS_FILE)
orders_items = pd.read_csv(ORDER_ITEMS_FILE)
payments_file = pd.read_csv(PAYMENTS_FILE)
reviews_file = pd.read_csv(REVIEWS_FILE)
customers_file = pd.read_csv(CUSTOMERS_FILE)
products_file = pd.read_csv(PRODUCTS_FILE)
sellers_file = pd.read_csv(SELLERS_FILE)
geolocation_file = pd.read_csv(GEOLOCATION_FILE)
category_translation_file = pd.read_csv(CATEGORY_TRANSLATION_FILE)
```

```python
evaluate_completeness(orders_items)
```

`order_items` has no missing values in any column, meaning every line item has a valid order, product, seller, shipping date, price, and freight value.

```python
evaluate_completeness(payments_file)
```

`payments` is fully complete across all columns, so every payment record has a valid order, type, installment count, and value.

```python
evaluate_completeness(reviews_file)
```

`reviews` is fully complete for structural fields (ids, score, dates), but the free-text fields are largely empty: `review_comment_title` is missing 88.34% of the time and `review_comment_message` 58.70%. This reflects that most customers leave a numeric score without writing a comment, so any text-based analysis on this table will only cover a minority of reviews.

```python
evaluate_completeness(customers_file)
```

`customers` has no missing values, so every customer record carries a complete id, zip code, city, and state.

```python
evaluate_completeness(products_file)
```

`products` has two small clusters of missing data: the descriptive attributes (`product_category_name`, name/description length, photos qty) are missing in 1.85% of rows likely products cataloged without full metadata while the physical dimensions (weight, length, height, width) are missing in only 2 rows, a negligible edge case.

```python
evaluate_completeness(sellers_file)
```

`sellers` is fully complete, with every seller record containing a valid id, zip code, city, and state.

```python
evaluate_completeness(geolocation_file)
```

`geolocation` is fully complete across its ~1M rows, with no missing zip codes, coordinates, city, or state values.

```python
evaluate_completeness(category_translation_file)
```

`category_translation` is fully complete, with all 71 categories mapped to their English equivalent.

# **Uniqueness**
---

```python
evaluate_uniqueness(orders)
```

`order_id` and `customer_id` are fully unique, confirming `orders` has one row per order and each order has its own customer record. Low-cardinality columns like `order_status` (8 values) and `order_estimated_delivery_date` (459 distinct dates) show expected repetition rather than a data quality issue.

```python
evaluate_uniqueness(orders_items)
```

`order_id` repeats across 87.59% of rows because orders can contain multiple items, which is the expected one-to-many relationship with `orders`. `product_id` and `seller_id` also repeat naturally, as the same product or seller appears in many line items.

```python
evaluate_uniqueness(payments_file)
```

`order_id` is duplicated in 4.28% of rows, consistent with orders that were split across multiple payment methods or installments (`payment_sequential` up to 29). `payment_type` has only 5 distinct values, as expected for a categorical field.

```python
evaluate_uniqueness(reviews_file)
```

`review_id` is duplicated in 0.82% of rows and `order_id` in 0.56%, suggesting a small number of orders received more than one review. `review_score` behaves as expected with only 5 distinct values (1–5).

```python
evaluate_uniqueness(customers_file)
```

`customer_id` is fully unique (one row per order-level customer record), while `customer_unique_id` repeats in 3.36% of rows  this is the key signal that the same physical customer placed multiple orders under different `customer_id` values, which matters for any repeat-purchase analysis.

```python
evaluate_uniqueness(products_file)
```

`product_id` is fully unique, confirming one row per product. The remaining columns are low-cardinality descriptive attributes (74 categories, dimension/weight buckets), so their repetition is expected rather than a duplication problem.

```python
evaluate_uniqueness(sellers_file)
```

`seller_id` is fully unique. `seller_zip_code_prefix` repeats in 27.43% of rows, and `seller_city`/`seller_state` repeat even more, since many sellers share the same location — an expected geographic concentration.

```python
evaluate_uniqueness(geolocation_file)
```

No column in `geolocation` is unique, and `geolocation_zip_code_prefix` repeats in 98.10% of rows — this table stores many lat/lng points per zip code (a reference/lookup table), so heavy repetition is by design, not a defect.

```python
evaluate_uniqueness(category_translation_file)
```

Both columns are fully unique, confirming a clean 1-to-1 mapping between each Portuguese category name and its English translation, with no duplicate entries.

# **Distribution**
---

```python
evaluate_distribution(orders)
```

All columns in `orders` are treated as categorical (ids and timestamps), so no numeric stats apply. `order_status` is dominated by `delivered` (97.02%), meaning cancellations and other non-delivered states are a small minority of the dataset important to keep in mind when using this table for delivery-time analysis.

```python
evaluate_distribution(orders_items)
```

`price` ranges from 0.85 to 6,735.00 with a mean of 120.65 well above the median of 74.99, indicating a right-skewed distribution with high-value outliers. `freight_value` shows the same pattern (mean 19.99 vs. median 16.26). Most orders (mean `order_item_id` of 1.20) contain very few items.

```python
evaluate_distribution(payments_file)
```

`credit_card` dominates `payment_type` at 73.92%, making it by far the preferred payment method. `payment_value` is strongly right-skewed (mean 154.10 vs. median 100.00, max 13,664.08), and `payment_installments` averages under 3, though it can go up to 24.

```python
evaluate_distribution(reviews_file)
```

`review_score` has a mean of 4.09 and median of 5.0, showing customers skew heavily toward positive reviews. Combined with the low completeness of comment fields seen earlier, this suggests most reviews are quick, positive ratings without detailed written feedback.

```python
evaluate_distribution(customers_file)
```

`customer_state` is heavily concentrated in `SP` (41.98%), and `sao paulo` is also the top city (15.63%), confirming the well-known concentration of Brazilian e-commerce activity in São Paulo state. `customer_zip_code_prefix` spans the full national range (1,003–99,990).

```python
evaluate_distribution(products_file)
```

`cama_mesa_banho` (bed, bath & table) is the top category at 9.37% of products. `product_weight_g` is extremely right-skewed (mean 2,276.47 vs. median 700, max 40,425), meaning a few heavy items pull the average well above typical products. Physical dimensions (length, height, width) show more moderate spread.

```python
evaluate_distribution(sellers_file)
```

`seller_state` is even more concentrated than customers, with `SP` accounting for 59.74% of sellers, and `sao paulo` as the top city (22.42%)  sellers are more geographically concentrated than buyers, likely reflecting where sourcing/logistics hubs are located.

```python
evaluate_distribution(geolocation_file)
```

Latitude and longitude ranges (roughly -36 to 45, and -101 to 121) extend beyond Brazil's actual borders, suggesting a small number of coordinate outliers or data errors that should be filtered before using this table for geospatial analysis. `SP` again leads as the top state (40.42%), consistent with the other tables.

```python
evaluate_distribution(category_translation_file)
```

With 71 categories evenly represented (top category at only 1.41%), `category_translation` is a flat reference table with no dominant class, as expected for a lookup/mapping table.
